## Simple Intelligent Agent

This simple agent receives a room condition as a percept and produces an action.

The agent follows a condition-action rule:

- If the room is dirty, the agent cleans.
- If the room is clean, the agent rests.
- If the room condition is unknown, the agent inspects.|


```python
def simple_reflex_agent(room_condition):
    """
    A simple reflex agent that decides an action based on the room condition.
    """
    room_condition = room_condition.strip().lower()

    if room_condition == "dirty":
        return "CLEAN"
    elif room_condition == "clean":
        return "REST"
    else:
        return "INSPECT"
```

```python
percepts = ["dirty", "clean", "dirty", "unknown"]

print("Simple Reflex Agent Test")
print("-" * 30)

for i, percept in enumerate(percepts, start=1):
    action = simple_reflex_agent(percept)
    print(f"Percept {i}: {percept:<8} -> Action: {action}")
```

## Performance Measure

A simple performance measure gives a score when the agent chooses the correct
action for the given percept.

```python
score = 0
total = len(percepts)

for percept in percepts:
 action = simple_reflex_agent (percept)

 if percept == "dirty" and action == "CLEAN":
    score += 1
 elif percept == "clean" and action == "REST":
    score += 1
 elif percept == "unknown" and action == "INSPECT":
    score += 1

print(f"Agent score: {score}/{total}")
```

## Guide Questions and Answers

### 1. What is an intelligent agent?
** Answer :** An intelligent agent is anything that can perceive its environment through sensors and act upon that environment through actuators, in order to achieve specific goals. It takes in information (percepts), processes it based on some internal logic or rules, and produces an action in response. Agents can range from simple (like the reflex agent above, which just maps conditions directly to actions) to complex (agents that learn, plan, or reason about long-term goals).

### 2. What are the sensors and actuators in this simple agent example?
** Answer :** In this example, the sensor input is the room condition percept.
The actuator output is the action such as CLEAN, REST, or INSPECT.
